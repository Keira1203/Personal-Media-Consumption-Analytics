```python
from lxml import etree
import pandas as pd

parser = etree.XMLParser(recover=True)
tree = etree.parse('../data/raw/export.xml', parser)
root = tree.getroot()

records = []
for record in root.iter('Record'):
    records.append(record.attrib)

df = pd.DataFrame(records)
print(f"finish！{len(df):,}records")
df.to_csv('health_data.csv', index=False)
df.head()
```

```python
import pandas as pd

df = pd.read_csv('health_data.csv')

sleep_df = df[df['type'] == 'HKCategoryTypeIdentifierSleepAnalysis'].copy()

sleep_df['startDate'] = pd.to_datetime(sleep_df['startDate'])

jan2026_sleep = sleep_df[
    (sleep_df['startDate'].dt.year == 2026) & 
    (sleep_df['startDate'].dt.month == 1)
].copy()

print(f"sleeping data of January: {len(jan2026_sleep):,}cases")
print(jan2026_sleep.head())
```

```python
import os

jan2026_sleep.to_csv('../data/processed/sleep_jan2026.csv', index=False)
```
